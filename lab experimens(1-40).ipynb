6. Perform basic video processing operations on the captured video
• Read captured video in python and display the video, in slow motion and in fast motion.


```python
import cv2
from google.colab import files

# Upload video
print("Upload Video")
uploaded = files.upload()
video_path = list(uploaded.keys())[0]

cap = cv2.VideoCapture(video_path)

fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Slow Motion Output (lower fps)
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
slow_out = cv2.VideoWriter("slow_motion.mp4", fourcc, fps/2, (width, height))

# Fast Motion Output (higher fps)
fast_out = cv2.VideoWriter("fast_motion.mp4", fourcc, fps*2, (width, height))

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    slow_out.write(frame)
    fast_out.write(frame)

cap.release()
slow_out.release()
fast_out.release()

print("✅ Slow Motion and Fast Motion Videos Saved")

# Download outputs
files.download("slow_motion.mp4")
files.download("fast_motion.mp4")
```

9. Perform Rotation of an image to clockwise and counter clockwise direction.

```python
from google.colab import files
from IPython.display import display
from PIL import Image
import numpy as np
import cv2
uploaded = files.upload()
file_name = list(uploaded.keys())[0]
image = Image.open(file_name)
display(image)
image_cv = cv2.cvtColor(np.array(image), cv2.COLOR_RGB2BGR)
h, w = image_cv.shape[:2]
center = (w // 2, h // 2)
clockwise = cv2.getRotationMatrix2D(center, -45, 1.0)
counter_clockwise = cv2.getRotationMatrix2D(center, 45, 1.0)
rot_cw = cv2.warpAffine(image_cv, clockwise, (w, h))
rot_ccw = cv2.warpAffine(image_cv, counter_clockwise, (w, h))
rot_cw_pil = Image.fromarray(cv2.cvtColor(rot_cw, cv2.COLOR_BGR2RGB))
rot_ccw_pil = Image.fromarray(cv2.cvtColor(rot_ccw, cv2.COLOR_BGR2RGB))
display(rot_cw_pil)
display(rot_ccw_pil)
```

12. Perform Perspective Transformation on the image.

```python
from google.colab import files
from IPython.display import display
from PIL import Image
import numpy as np
import cv2
uploaded = files.upload()
file_name = list(uploaded.keys())[0]
image = Image.open(file_name)
display(image)
image_cv = cv2.cvtColor(np.array(image), cv2.COLOR_RGB2BGR)
h, w = image_cv.shape[:2]
pts1 = np.float32([[50, 50], [w-50, 50], [50, h-50], [w-50, h-50]])
pts2 = np.float32([[0, 0], [w, 100], [100, h], [w, h]])
M = cv2.getPerspectiveTransform(pts1, pts2)
perspective = cv2.warpPerspective(image_cv, M, (w, h))
perspective_pil = Image.fromarray(cv2.cvtColor(perspective, cv2.COLOR_BGR2RGB))
display(perspective_pil)
```

13. Perform Perspective Transformation on the Video.

```python
import cv2
import numpy as np
from google.colab import files

# Upload video
print("Upload Video")
uploaded = files.upload()
video_path = list(uploaded.keys())[0]

cap = cv2.VideoCapture(video_path)

fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter("perspective_output.mp4", fourcc, fps, (width, height))

# Perspective points (Change if needed)
pts1 = np.float32([[50,50],[width-50,50],[50,height-50],[width-50,height-50]])
pts2 = np.float32([[0,0],[width,0],[100,height],[width-100,height]])

M = cv2.getPerspectiveTransform(pts1, pts2)

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    perspective = cv2.warpPerspective(frame, M, (width, height))
    out.write(perspective)

cap.release()
out.release()

print("✅ Perspective Video Saved")

# Download output
files.download("perspective_output.mp4")
```

36. Recognise watch from the given image by general Object recognition using OpenCV.

```python
import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow
uploaded = files.upload()
img_name = list(uploaded.keys())[0]
img = cv2.imread(img_name)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
blur = cv2.GaussianBlur(gray, (5, 5), 0)
edges = cv2.Canny(blur, 50, 150)
contours, _ = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
for cnt in contours:
    area = cv2.contourArea(cnt)
    if area > 2000:
        x, y, w, h = cv2.boundingRect(cnt)
        if 0.8 < w/h < 1.2:
            cv2.rectangle(img, (x, y), (x+w, y+h), (0, 255, 0), 2)
            cv2.putText(img, "Watch Detected", (x, y-10),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,255,0), 2)
cv2_imshow(img)
```

```python
from google.colab import drive
drive.mount('/content/drive')
import cv2
from google.colab import files

# Upload video
uploaded = files.upload()
video_path = list(uploaded.keys())[0]

cap = cv2.VideoCapture(video_path)

frames = []

# Read all frames
while True:
    ret, frame = cap.read()
    if not ret:
        break
    frames.append(frame)

# Get video properties
h, w = frames[0].shape[:2]
fps = cap.get(cv2.CAP_PROP_FPS)
cap.release()

# Video writer (output file)
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter("output.mp4", fourcc, fps, (w, h))

# Write frames in reverse
for frame in reversed(frames):
    out.write(frame)

out.release()

print("Saved as output.mp4")
```

38.Face Detection using Opencv.

```python
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow
uploaded = files.upload()
img_name = list(uploaded.keys())[0]
img = cv2.imread(img_name)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)
faces = face_cascade.detectMultiScale(gray, 1.3, 5)
for (x, y, w, h) in faces:
    cv2.rectangle(img, (x, y), (x+w, y+h), (0, 255, 0), 2)
cv2_imshow(img)
```

39. Vehicle Detection in a Video frame using OpenCV .


```python
import cv2
import numpy as np
from google.colab import files

# Upload video
print("Upload Traffic Video")
uploaded = files.upload()
video_path = list(uploaded.keys())[0]

cap = cv2.VideoCapture(video_path)

fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter("vehicle_output.mp4", fourcc, fps, (width, height))

# Background Subtractor
fgbg = cv2.createBackgroundSubtractorMOG2()

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    fgmask = fgbg.apply(frame)

    # Remove noise
    kernel = np.ones((5,5), np.uint8)
    fgmask = cv2.morphologyEx(fgmask, cv2.MORPH_CLOSE, kernel)

    # Find contours
    contours, _ = cv2.findContours(fgmask, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)

    for cnt in contours:
        if cv2.contourArea(cnt) > 1500:   # Ignore small objects
            x, y, w, h = cv2.boundingRect(cnt)
            cv2.rectangle(frame, (x,y), (x+w,y+h), (0,255,0), 2)

    out.write(frame)

cap.release()
out.release()

print("✅ Vehicle Detection Video Saved")

files.download("vehicle_output.mp4")
```
